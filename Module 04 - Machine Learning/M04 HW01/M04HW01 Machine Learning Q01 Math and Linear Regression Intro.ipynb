## Task 1
Draw the lines for:
> x + y = 12
> x - y = 4

Add grid, legend, and axis lines

```python
x_values = np.arange(-20, 21, 10)

y_values1 = 12 - x_values
y_values2 = x_values - 4

plt.axhline(0, color='darkgray')
plt.axvline(0, color='darkgray')

plt.plot(x_values, y_values1, color='orange', label='x + y = 12')
plt.plot(x_values, y_values2, color='purple', label='x - y = 4')

plt.grid(color='gray', ls='--', lw=0.5)
plt.legend()

plt.show()
```

## Task 2
Use symbols, Eq, and solve to find the intersection point. Scatter the point and annotate it on the graph

```python
x = symbols('x')
eq1 = 12 - x
eq2 = x - 4

solution = solve(Eq(eq1, eq2), x)

x_intersect = float(solution[0])
y_intersect = x_intersect - 4

x_values = np.arange(-20, 21, 10)

y_values1 = 12 - x_values
y_values2 = x_values - 4

plt.axhline(0, color='darkgray')
plt.axvline(0, color='darkgray')

plt.plot(x_values, y_values1, color='orange', label='x + y = 12')
plt.plot(x_values, y_values2, color='purple', label='x - y = 4')

plt.grid(color='gray', ls='--', lw=0.5)
plt.legend()

plt.scatter(x_intersect, y_intersect, color='red', zorder=10)
plt.annotate(
    f'({int(x_intersect)},{int(y_intersect)})',
    xy=(x_intersect, y_intersect),
    xytext=(4, 12),
    arrowprops=dict()
)

plt.show()
```
