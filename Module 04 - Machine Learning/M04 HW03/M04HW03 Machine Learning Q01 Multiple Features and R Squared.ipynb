# Question 1
Compute 
R<sup>2</sup> by hand (no sklearn) for this tiny dataset.
Find $\bar{y}$, then 
SST, then 
SSE, then R<sup>2</sup>. Show every step.

> y      = np.array([10, 12, 20, 18])  
> y_pred = np.array([11, 13, 18, 18])

````python
y = np.array([10, 12, 20, 18])
y_pred = np.array([11, 13, 18, 18])

SST = np.sum((y - np.mean(y)) ** 2)
SSE = np.sum((y - y_pred) ** 2)

R2 = 1 - SSE / SST

print('mean =', np.mean(y))
print('SST =', SST)
print('SSE =', SSE)
print('R2 =', R2)
````

# Question 3
Use the following dataset. X contains 3 features, and y contains the matching target values.
```
X = np.array([  
    [2, 1, 3], [4, 2, 1], [6, 3, 5], [8, 1, 4],  
    [3, 5, 2], [5, 4, 6], [7, 2, 3], [9, 5, 1],  
    [1, 3, 4], [10, 4, 5]  
])  
```
y = np.array([18, 20, 36, 40, 15, 32, 33, 30, 17, 43])  
1. Use LinearRegression to find and print the intercept and coefficients
2. Find the MSE
3. Find the R² score using r2_score
4. Explain whether the R² value you found is a good score or not

````python
X = np.array([  
    [2, 1, 3], [4, 2, 1], [6, 3, 5], [8, 1, 4],  
    [3, 5, 2], [5, 4, 6], [7, 2, 3], [9, 5, 1],  
    [1, 3, 4], [10, 4, 5]  
]) 
y = np.array([18, 20, 36, 40, 15, 32, 33, 30, 17, 43])

model = LinearRegression()
model.fit(X, y)

print('coefficients:', model.coef_)
print('intercept:', model.intercept_)

y_pred = model.predict(X)
mse = mean_squared_error(y, y_pred)
print('MSE:', mse)

R2 = r2_score(y, y_pred)
print('R\N{SUPERSCRIPT TWO}:', R2)

# the r2 value fopund is very good. the closer to 1 it is, the better, since it means the model can properly explain the variance
````
