1. Eliminate any row in the DataFrame that contains at least one NaN value
2. Modify the previous task and eliminate only rows that have NaN values across all columns
3. Eliminate rows that have at least one NaN in columns A and C only
4. Replace all NaN values with a specified value of your choice

```python
df_missing = pd.DataFrame({
'A': [None, 2, 3, None, 5, None],
'B': [10, None, None, 40, 50, None],
'C': [100, None, 300, 400, None, None]
})

# 1
df_missing.dropna()

# 2
df_missing.dropna(thresh=1)

# 3
df_missing.dropna(subset=['A', 'C'])

# 4
df_missing.fillna('Chocolate')

# 5
df_missing['B'].fillna(df_missing['B'].mean())
```
