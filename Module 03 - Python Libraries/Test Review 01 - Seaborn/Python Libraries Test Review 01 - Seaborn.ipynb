```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('mpg.csv')
```

# Exercise 1
## Scatterplot

#### Scatter plot: horsepower vs mpg, colored by origin
Create a scatter plot (`scatterplot`) between the `horsepower` column and the `mpg` column.  
Add different colors by the `origin` column (car origin) using the `hue` parameter

![Result Graph](python_libs_test_review_p1q1_graph.jpeg)

```python
scatter_df = df[df['horsepower'].astype(str).str.isdigit()].copy()
scatter_df['horsepower'] = scatter_df['horsepower'].astype(int)
scatter_df['origin'] = scatter_df['origin'].replace([1, 2, 3], ['USA', 'Europe', 'Japan'])

sns.scatterplot(data=scatter_df,
                x='horsepower',
                y='mpg',
                hue='origin',
                palette='bright')
plt.style.use("dark_background")
plt.title('horsepower vs mpg')
plt.xlabel('horsepower')
plt.ylabel('mpg')
plt.show()
```

# Exercise 2
## Barplot


#### Do 4-cylinder cars use less fuel than 8-cylinder cars?
Check whether cars with 4 cylinders consume less fuel than cars with 8 cylinders (using the `mpg` column).  
Show it with a bar chart (`barplot`). Higher `mpg` means the car uses less fuel

![Result Graph](python_libs_test_review_p1q2_graph.jpeg)

```python
barplot_df = df[df['cylinders'].isin([4, 8])].copy()
ax = sns.barplot(
    data=barplot_df,
    x='cylinders',
    y='mpg',
    hue='cylinders',
    palette='bright',
    errorbar=None
)
plt.style.use("dark_background")
plt.bar_label(ax.containers[0], fmt='%.0f~')
plt.bar_label(ax.containers[1], fmt='%.0f~')
plt.title('Mean mpg by cylinders (4 vs 8)')
plt.ylabel('mpg')
plt.xlabel('cylinders')

plt.show()
```

# Exercise 3
## Countplot


#### Count plot: which origin has the most cars?
Create a count plot (`countplot`) for the `origin` column to check which country contributes the most cars to the dataset.  
A count plot automatically counts the rows for each category

![Result Graph](python_libs_test_review_p1q3_graph.jpeg)

```python
countplot_df = df.copy()
countplot_df['origin'] = countplot_df['origin'].replace([1, 2, 3], ['USA', 'Europe', 'Japan'])

ax = sns.countplot(
    data=countplot_df,
    x='origin',
    hue='origin',
    palette='bright'
)

plt.style.use("dark_background")
plt.bar_label(ax.containers[0], fmt='%.0f')
plt.bar_label(ax.containers[1], fmt='%.0f')
plt.bar_label(ax.containers[2], fmt='%.0f')
plt.title('count of cars per origin')
plt.ylabel('count')
plt.xlabel('')
plt.show()
```

# Exercise 4
## Histplot


#### Histogram of the weight column
Create a `histogram` for the `weight` column to show the distribution of car weights across the dataset.
A histogram groups the values into bins and counts how many cars fall into each bin

![Result Graph](python_libs_test_review_p1q4_graph.jpeg)

```python
histplot_df = df.copy()

ax = sns.displot(
    data = histplot_df,
    x = 'weight',
    bins = 20,
    color = '#7C4DFF',
    edgecolor = '#161822'
)

plt.title('distribution of weight')
plt.style.use("dark_background")
plt.show()
```

# Exercise 5
## Boxplot


#### Box plot: compare mpg by number of cylinders
Create a box plot (`boxplot`) to compare `mpg` values by the number of cylinders (`cylinders`).  
A box plot shows the median, the middle 50% (the box), and outliers for each group

![Result Graph](python_libs_test_review_p1q5_graph.jpeg)

```python
boxplot_df = df.copy()

ax = sns.boxplot(
    data=boxplot_df,
    x='cylinders',
    y='mpg',
    hue='cylinders',
    palette='bright'
)

plt.title('mpg distribution by cylinders')
plt.show()
```
