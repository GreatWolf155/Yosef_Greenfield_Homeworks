# Dataset Preparation

```python
csv_data = """order_id,customer,city,total_bill,tip,table_size
1,Alice,Tel Aviv,45.50,8.00,4
2,Bob,Haifa,22.30,3.50,2
3,Carol,Tel Aviv,67.80,12.00,6
4,David,Jerusalem,15.20,2.00,2
5,Eve,Haifa,89.90,15.00,8
6,Frank,Tel Aviv,33.40,5.50,3
7,Grace,Jerusalem,51.00,9.00,4
8,Henry,Haifa,28.60,4.00,2
9,Iris,Tel Aviv,74.20,13.50,5
10,Jake,Jerusalem,19.80,2.50,2"""
```

## Task 1
Load csv_data into a DataFrame using io.StringIO and pd.read_csv. Print .info() and .describe().transpose()

```python
df = pd.read_csv(io.StringIO(csv_data))
df
```

```python
print(df.info())
print()
print(df.describe().transpose())
```

## Task 2
Add a column price_per_person that holds the bill divided by the table size, rounded to 2 decimal places.

```python
def price_per_person(bill, table_size):
    return round(bill / table_size, 2)
df['price_per_person'] = np.vectorize(price_per_person)(df['total_bill'], df['table_size'])
df
```

## Task 3
Filter the DataFrame to show only orders where price_per_person is greater than 12

```python
df[df['price_per_person'] > 12]
```

## Task 4
Use .apply() to create a column table_type: 'Large' if table_size >= 4, otherwise 'Small'

```python
def small_or_large(table_size):
    if table_size >= 4:
        return 'Large'
    return 'Small'

df['table_type'] = df['table_size'].apply(small_or_large)
df
```

## Task 5
Set customer as the index. Then use .loc[] to select the rows from Alice to Carol

```python
df = df.set_index('customer')
df.loc['Alice':'Carol']
```

## Task 6
Use pd.concat with axis=0 to create a new DataFrame containing the first 3 rows and the last 3 rows of the original DataFrame. Reset the index

```python
new_df = pd.concat(
    [df.head(3), df.tail(3)],
    axis=0
    ).reset_index()
new_df
```

## Task 7
Use np.vectorize to create a column value_label:
'High Value' if total_bill > 50 and tip > 10%, otherwise 'Standard'

```python
def value_label(bill, tip):
    if (bill > 50) & ((tip / bill) > .1):
        return 'High Value'
    return 'Standard'
df['value_label'] = np.vectorize(value_label)(df['total_bill'], df['tip'])
df
```
