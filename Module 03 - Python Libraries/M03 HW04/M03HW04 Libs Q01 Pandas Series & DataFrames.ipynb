## Task 1

```python
workers = pd.Series(['Alice', 'Bob', 'Carol', 'David'])
mon_workers = pd.Series(data=[8, 6, 7], index=workers[workers != 'Carol'])
tues_workers = pd.Series(data=[8, 5, 6], index=workers[workers != 'Alice'])
wed_workers = pd.Series(data=[7, 8, 5], index=workers[workers != 'Bob'])
```

## Task 2

```python
total_hours = mon_workers.add(tues_workers, fill_value=0).add(wed_workers, fill_value=0).astype(int)
total_hours
```
