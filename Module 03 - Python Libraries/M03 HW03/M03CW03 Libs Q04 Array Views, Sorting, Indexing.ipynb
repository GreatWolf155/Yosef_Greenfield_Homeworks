# Question 3
Given a = np.array([1,2,3]) and v = a.view() — if you do v[0] = 99, what does a[0] return?  
view shares memory  
a[0] = ?

````python
a = np.array([1,2,3])
v = a.view()
v[0] = 99
a[0] # returns 99. view shows you the array, and changing what is in the view will therefore change what it is looking at
````

# Question 4
Given a = np.array([1,2,3]) and c = np.copy(a) — if you do c[0] = 99, what does a[0] return?  
copy is independent  
a[0] = ?

````python
a = np.array([1,2,3])
c = np.copy(a)
c[0] = 99
a[0] # returns 1, since copy does not view the original array, but copies it to a new one
````

# Question 7

array([?, ?])

````python
b = np.array([[1,2],[3,4],[5,6]])
b[[0,2], [1,0]] # same as a condition, will give [5, 2], since it moves first to the right and then down
````
