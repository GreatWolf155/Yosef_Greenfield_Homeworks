# Question 4
Given A = np.array([[1,2],[3,4]]) and B = np.array([[1,0],[0,1]]) — what does A.dot(B) return?  
B is the identity matrix
```
array([[?, ?],
       [?, ?]])
```

````python
A = np.array([[1,2],[3,4]])
B = np.array([[1,0],[0,1]])
A.dot(B) # returns the dot product of the two arrays, [[1, 2], [3, 4]]
````

### Dot Product Explanation
![Dot Product](Dot_Product.PNG)  
In our case, we have:
```
  ⌈ 1  2 ⌉       ⌈ 1  0 ⌉        ⌈ 1×1 + 2×0   1×0 + 2×1 ⌉       ⌈ 1  2 ⌉
            •              =                               =  
  ⌊ 3  4 ⌋       ⌊ 0  1 ⌋        ⌊ 3×1 + 4×0   3×0 + 4×1 ⌋       ⌊ 3  4 ⌋
```
