```python
np.random.seed(57)
arr = np.random.randint(0, 101, (6, 7))
arg_max = arr.argmax()

col = arg_max % arr.shape[1]
row = arg_max // arr.shape[1]

print(f'Coordinates of largest number: ({row},{col})')
```
