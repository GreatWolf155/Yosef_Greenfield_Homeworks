### Mean, Variance, and Standard Deviation Equations
![mean_variance_std.PNG](mean_variance_std.PNG)  
In our case, we have a sum of: 40  
And a len of: 8  
Meaning, the average, or mean, is: 40 / 8 = 5  
So now we have [2-5, 4-5, 4-5, 4-5, 5-5, 5-5, 7-5, 9-5] = [-3, -1, -1, -1, 0, 0, 2, 4]  
Square each of those: [9, 1, 1, 1, 0, 0, 4, 16]  
Add them up: 32  
Now we divide by the number of values, which is 8: 4  
The square root of that gives us the **Standard Deviation**: 2.0

# Question 6
Given x = np.array([1,2,3]) and y = np.array([2,4,6]) — what is the off-diagonal value of np.corrcoef(x, y)?  
perfectly proportional arrays  
r = ?.?

```python
x = np.array([1,2,3])
y = np.array([2,4,6])
np.corrcoef(x, y) # the diagonal is always 1.0 for all values across it, and the Pearson Correlation Coefficient will be explained below
```

### Pearson Correlation Coefficient
$$\begin{bmatrix} r_{xx} & r_{xy} \\ r_{yx} & r_{yy} \end{bmatrix}$$

$r_{xx}$ and $r_{yy}$ are always equal to 1, since the two arrays correlate perfectly with themselves  
$r_{xy}$ and $r_{yx}$ are the correlation between arrays x and y, and are equal  

We start by finding the mean for both arrays:  
$$\bar{X} = \frac{a + b + c}{3} = \frac{1 + 2 + 3}{3} = 2$$  
$$\bar{Y} = \frac{d + e + f}{3} = \frac{4 + 5 + 6}{3} = 5$$  

Next we find the correlation of x with y:  
$$r_{xy} = \frac{(a - \bar{X})(d - \bar{Y}) + (b - \bar{X})(e - \bar{Y}) + (c - \bar{X})(f - \bar{Y})}{\sqrt{[(a - \bar{X})^2 + (b - \bar{X})^2 + (c - \bar{X})^2] \cdot [(d - \bar{Y})^2 + (e - \bar{Y})^2 + (f - \bar{Y})^2]}}$$  
Where in our case looks like:  
$$r_{xy} = \frac{(1 - 2)(4 - 5) + (2 - 2)(5 - 5) + (3 - 2)(6 - 5)}{\sqrt{[(1 - 2)^2 + (2 - 2)^2 + (3 - 2)^2] \cdot [(4 - 5)^2 + (5 - 5)^2 + (6 - 5)^2]}} = \frac{1 + 0 + 1}{\sqrt{[(-1)^2 + 0^2 + 1^2] \cdot [(-1)^2 + 0^2 + 1^2]}} = \frac{2}{\sqrt{4}} = 1$$  

It looks complicated, but its really just plugging in the numbers, and you get an answer; the answer here being that they have a correlation coefficient of 1 with each other, meaning the output matrix will be:  
$$\begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix}$$
