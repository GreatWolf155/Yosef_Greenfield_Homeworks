# Question 4
Given a = np.arange(8).reshape(4,2) — what does np.vsplit(a, 2) return?  
splits into 2 sub-arrays of shape (2, 2)  
[array([[?,?],[?,?]]), array([[?,?],[?,?]])]

````python
a = np.arange(8).reshape(4,2)
np.vsplit(a, 2) # splits array a into 2 arrays. goes in order of numbers - vertical split, like argmin and argmax.
# basically takes the rows and splits them based on the number of times requested
# must be a divisor of the total n umber of rows
# returns [[0, 1], [2, 3]], [[4, 5], [6, 7]]
````

# Question 5
Given a = np.array([1,2,3,4,5,6]) — what does np.hsplit(a, 3) return?  
splits into 3 equal parts  
[array([?,?]), array([?,?]), array([?,?])]

````python
a = np.array([1,2,3,4,5,6])
np.hsplit(a, 3) # similar to vsplit, only horizontally, meaning it will split the columns
# returns [1, 2], [3, 4], [5, 6]
````

# Question 12
Given a = np.array([1,2,3]) and b = np.array([4,5,6]) — what does np.concatenate([a, b]) return?  
array([?, ?, ?, ?, ?, ?])

````python
a = np.array([1,2,3])
b = np.array([4,5,6])
np.concatenate([a, b])
# while vstack and hstack can create more dimensions automatically, concatenate requires the dimensoin to already exist
# np.concatenate([a, b], axis = 1), while vstack will return a 2 dimensional array
# returns [1, 2, 3, 4, 5, 6]
````

# Question 13
Given a = np.array([1,2,3]) and b = np.array([4,5,6]) — what does np.column_stack([a, b]) return?  
shape: (3, 2)  
```
array([[?, ?],
       [?, ?],
       [?, ?]])
```

````python
a = np.array([1,2,3])
b = np.array([4,5,6])
np.column_stack([a, b])
# if the arrays given are both multidimensional, this will effectively do the same as transpose and then hstack
# transpose does not do anything to single dimensional arrays, so single dimensional arrays for that method will return a one dimensional array

# interestingly, vstack and then transpose WILL actually give you the same result as column_stack, although in a different manner
# vstack creates a view, while column_stack creates a brand new array

# will return [[1, 4], [2, 5], [3, 6]]
````
