# 📝 Tasks

Open Jupyter Notebook and create a new notebook.
Change the first cell type to Markdown (press M in Command Mode).
Write # Your Name as a level-1 heading.
Add ## About Me and write one sentence about yourself in italic using *text*.
Add ## Hobbies and list at least 3 real hobbies as a numbered list.
Run the cell with Shift + Enter — check that it looks like the preview above.
Save your notebook (Ctrl + S), upload it to GitHub, and send the link to the email below.
