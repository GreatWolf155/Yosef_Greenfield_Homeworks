## Task 1
Load tips.csv. Display the first rows, df.info(), and df.describe().T  
Then add a new column price_per_person calculated as total_bill / size.

```python
tips_df = pd.read_csv('tips.csv')
df = tips_df.copy()

display(df.head())
display(df.info())
display(df.describe().T)

def p_per_p(total_bill, size):
    return (total_bill / size)
    
df['price_per_person'] = np.vectorize(p_per_p)(
    df['total_bill'],
    df['size']
)
df
```

## Task 2
Count how many orders (rows) were placed at each time and for each gender.

```python
df.groupby(['time', 'sex'])['total_bill'].count().rename('total_orders')
```

## Task 3
Count how many orders (rows) were placed on each day.

```python
df.groupby(['day'])['total_bill'].count().rename('total_orders_per_day')
```

## Task 4
Find the maximum price_per_person for each day and gender combination.

```python
df.groupby(['day', 'sex'])['price_per_person'].max()
```

## Task 5
Find the minimum price_per_person for each day and size combination.

```python
df.groupby(['day', 'size'])['price_per_person'].min()
```

## Task 6
Calculate the sum of total_bill grouped by size and smoker status.  
Then use .loc to extract the sum for size 2 smokers only.

```python
df.groupby(['size', 'smoker'])['total_bill'].sum()

df.groupby(['size', 'smoker'])['total_bill'].sum().loc[(2, 'Yes')]
```

## Task 7
Display the maximum price_per_person grouped by day and gender  
Then use .loc to show the results for Male only across all days.

```python
df.groupby(['day', 'sex'])['price_per_person'].max()

df.groupby(['sex', 'day'])['price_per_person'].max().loc[('Male')]

# alternative answer
df.groupby(['day', 'sex'])['price_per_person'].max().xs('Male', level='sex')
```
