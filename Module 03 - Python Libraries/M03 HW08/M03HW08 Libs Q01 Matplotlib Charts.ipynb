## Task 1
Create a line chart representing the trend of GDP per Capita over the years for a country of your choice.  
Ensure the line chart has a title and labelled axes.

```python
df = pd.read_csv('matplotlib_charts_class_exc.csv')
df_USA_GDP = df[df['Country'] == 'USA']

x = df_USA_GDP['Year']
y = df_USA_GDP['GDP per Capita']
plt.plot(x, y, color='red')

plt.title("USA GDP per Capita by Year")
plt.xlabel("Year")
plt.ylabel("GDP per Capita")
plt.legend(["This line is red to make it look more official"])

plt.show()
```

## Task 2
Using a histogram, visualize the distribution of the GDP per Capita values for a particular year of your choice.

```python
y = df[df['Year'] == 2001]['GDP per Capita'] # honestly surprised this one worked on the first try

plt.hist(
    y,
    bins=df['Country'].nunique(),
    color="skyblue",
    edgecolor="black"
)

plt.title("GDPpC Distribution")
plt.xlabel("GDPs")
plt.ylabel("Number of Countries")
plt.tight_layout()
plt.show()
```

## Task 3
Plot a bar chart to compare the GDP per Capita for five different countries in the year 2010.

```python
countries = df['Country'].drop_duplicates().sample(n=5)

df2010 = df[df['Year'] == 2010]
df_sample = df2010.sample(n=5) # should get 5 different countries

countries = df_sample['Country']
GDP = df_sample['GDP per Capita']


bar_colors = ["skyblue", "gold", "lightgreen", "tomato", "blue"]

plt.bar(
    countries,
    GDP,
    color=bar_colors,
    edgecolor="black"
)
plt.ylabel("GDP per Capita")
plt.xlabel("Country")
plt.title("Countries")
plt.tight_layout()
plt.show()
```
